```python
import numpy as np
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from datetime import date

time = date.today()
str_time = f"{time.year}-{time.month}-{time.day}"
str_time
```

```python
df = yf.download("BTC-USD" , start="2018-01-01" , end=str_time , multi_level_index=False)
df.head()
```

```python
plt.figure(figsize=(16 , 6))
df["Close"].plot(kind="line")
plt.show()
```

```python
plt.figure(figsize=(16 , 6))

plt.plot(df["Close"])

df["SMA_30"] = df["Close"].rolling(window=30).mean()

plt.plot(df["SMA_30"])

plt.show()
```

```python
plt.figure(figsize=(16 , 6))

plt.plot(df["Close"])

df["EMA_7"] = df["Close"].ewm(span=7).mean()

plt.plot(df["EMA_7"])

plt.show()
```

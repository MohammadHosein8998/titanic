## importing packages

```python
import numpy as np
import seaborn as sb
import matplotlib.pyplot as plt
import pandas as pd
```

## loading the titanic data

```python
df = sb.load_dataset("titanic")
df.head(10)
```

## تحلیل اکتشافی دیتا

### هیستوگرام نجات یافته ها و کشته شده ها 

```python
plt.hist(df["alive"] , bins=2)
plt.show()
```

### هیستوگرام نجات یافته ها و کشته شده ها برای مرد ها 

```python
plt.hist(df[df["who"] == "man"]["alive"] , bins=2)
plt.show()
```

```python
for class_num in ["First" , "Second" , "Third"]:
        plt.hist(df[(df["who"] == "man") & (df["class"] == class_num) & (df["alone"] == alone)]["alive"] ,
                 bins=2 , label=f"man-{class_num}")
        plt.legend()
        plt.show()
```

### هیستوگرام نجات یافته ها و کشته شده ها در شهر های مختلف

```python
for town in df["embark_town"].unique():
    plt.hist(df[df["embark_town"] == str(town)]["alive"] , bins=2 , label=f"alive-{town}")
    plt.legend()
    plt.show()
```

### هیستوگرام نجات یافته ها و کشته شده ها برای خانم ها

```python
for class_num in ["First" , "Second" , "Third"]:
        plt.hist(df[(df["who"] == "woman") & (df["class"] == class_num) & (df["alone"] == alone)]["alive"] ,
                 bins=2 , label=f"woman-{class_num}")
        plt.legend()
        plt.show()
```

### هیستوگرام نجات یافته ها و کشته شده ها برای بچه ها

```python
plt.hist(df[df["who"] == "child"]["alive"] , bins=2 )

plt.show()
```
